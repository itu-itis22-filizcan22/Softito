```python
def sayi_durumu(sayi):
    if sayi > 0:
        return "Pozitif"
    elif sayi < 0:
        return "Negatif"
    else:
        return "Sıfır"

def en_buyugu_bul(a, b, c):
    en_buyuk = a
    if b > en_buyuk:
        en_buyuk = b
    if c > en_buyuk:
        en_buyuk = c
    return en_buyuk

print(f"-5 sayısı durumu: {sayi_durumu(-5)}")
print(f"45, 12, 67 arasından en büyüğü: {en_buyugu_bul(45, 12, 67)}")
```

```python
# Boş liste oluşturup 5 şehir alma
sehirler = []
for i in range(5):
    yeni_sehir = input(f"{i+1}. Şehri giriniz: ")
    sehirler.append(yeni_sehir)


print("\n--- Girdiğiniz Şehirler ---")
for index, sehir in enumerate(sehirler, 1):
    print(f"{index}- {sehir}")


silinecek = input("\nListeden silmek istediğiniz şehri yazın: ")
if silinecek in sehirler:
    sehirler.remove(silinecek)
    print(f"{silinecek} silindi. Güncel liste: {sehirler}")
else:
    print("Bu şehir listede yok!")
```

```python
class Araba:
    def __init__(self, marka, model, hiz=0):
        self.marka = marka
        self.model = model
        self.hiz = hiz

    def hizlandir(self, artis):
        self.hiz += artis
        print(f"Araç {artis} km/s hızlandı. Güncel hız: {self.hiz} km/s")

    def bilgi_goster(self):
        print(f"Marka: {self.marka} | Model: {self.model} | Hız: {self.hiz} km/s")


araba1 = Araba("Volvo", "EX40", 50)


araba1.bilgi_goster()


araba1.hizlandir(30)


araba1.bilgi_goster()
```

```python
class dikdortgen :
  def __init__(self,uzunluk,genislik) :
    self.uzunluk = uzunluk
    self.genislik = genislik

def alan_hesapla(self) :
  return self.uzunluk * self.genislik

def cevre_hesapla(self) :
  return self.uzunluk * 2 + self.genislik * 2
```

```python
class Kitap:
    def __init__(self, ad, yazar, sayfa_sayisi):
        self.ad = ad
        self.yazar = yazar
        self.sayfa_sayisi = sayfa_sayisi

    def ozet(self):
        return f"{self.ad} - {self.yazar} ({self.sayfa_sayisi} sayfa)"


kitap1 = Kitap("Suç ve Ceza", "Dostoyevski", 671)

print(kitap1.ozet())
```
