```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Grafiklerin görsel stilini standart hale getirme
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (10, 6) # Varsayılan grafik boyutu (Genişlik x Yükseklik)

# randomness sabitlenmesi
np.random.seed(42)


pd.set_option('display.max_columns', None) # Tüm sütunları göster
pd.set_option('display.max_rows', 100)     # 100 satıra kadar kısaltmadan göster
pd.set_option('display.float_format', lambda x: '%.2f' % x) # Ondalık sayıları 2 basamak göster
```

```python
n_samples = 1000

sehir_havuzu = ['İstanbul', 'istanbul ', ' Ankara', 'ANKARA', 'İzmir', 'izmir', 'Bursa']


df = pd.DataFrame({
    'yas': np.random.normal(loc=35, scale=10, size=n_samples).astype(int),
    'gelir': np.random.normal(loc=15000, scale=5000, size=n_samples),
    'sehir': np.random.choice(sehir_havuzu, size=n_samples),
    'abonelik_tipi': np.random.choice(['Aylık', 'Yıllık', 'Premium'], size=n_samples, p=[0.5, 0.3, 0.2]),
    'kayit_tarihi': pd.date_range(start='2022-01-01', periods=n_samples, freq='D'),
    'aylik_harcama': np.random.uniform(50, 500, size=n_samples),
    'memnuniyet_puani': np.random.randint(1, 6, size=n_samples),
    'churn': np.random.choice([0, 1], size=n_samples, p=[0.8, 0.2]) # %20 terk oranı
})
```

```python
# Eksik değerler ekleme
df.loc[df.sample(frac=0.05).index, 'gelir'] = np.nan
df.loc[df.sample(frac=0.03).index, 'yas'] = np.nan
df.loc[df.sample(frac=0.10).index, 'memnuniyet_puani'] = np.nan

# Gelire  Outlier ekleme
df.loc[np.random.choice(df.index, size=5), 'gelir'] = np.random.uniform(150000, 300000, size=5)

# Yaşa mantıksız değerler ekleme
df.loc[np.random.choice(df.index, size=3), 'yas'] = [-5, 150, 200]

# Duplicate satırlar ekleme
kopya_satirlar = df.sample(n=20, random_state=42)
df = pd.concat([df, kopya_satirlar], ignore_index=True)
```

```python
# Satır ve sütun sayısı
print(f"Veri Seti Boyutu: {df.shape[0]} satır, {df.shape[1]} sütun\n")


print("--- Veri Tipleri ve Dolu Hücreler ---")
df.info()


display(df.head())    # İlk 5 satır
display(df.tail())    # Son 5 satır
display(df.sample(5)) # Rastgele 5 satır
```

```python
eksik_deger_tablosu = pd.DataFrame({
    'Eksik Sayısı': df.isnull().sum(),
    'Eksik Yüzdesi (%)': (df.isnull().sum() / len(df)) * 100,
    'Benzersiz Değer Sayısı': df.nunique()
})
# Sadece eksik değeri olan sütunları sırala
display(eksik_deger_tablosu[eksik_deger_tablosu['Eksik Sayısı'] > 0].sort_values(by='Eksik Sayısı', ascending=False))

# Eksik değerlerin ısı haritası
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title("Eksik Değerlerin Veri Setindeki Dağılımı")
plt.show()
```

```python
tekrar_sayisi = df.duplicated().sum()
print(f"Veri setindeki yinelenen satır sayısı: {tekrar_sayisi}")
```

```python
# Tutarsız yazımların dağılımını görme
print("--- Temizlik Öncesi Şehir Dağılımı ---")
print(df['sehir'].value_counts())

# Şehir isimleri küçük harf, boşlukları silme
df['sehir'] = df['sehir'].str.lower().str.strip()

print("\n--- Temizlik Sonrası Şehir Dağılımı ---")
print(df['sehir'].value_counts())
```

```python
mantiksiz_yaslar = df[(df['yas'] < 0) | (df['yas'] > 100)]
print(f"Mantıksız yaş değerine sahip kayıt sayısı: {len(mantiksiz_yaslar)}")
display(mantiksiz_yaslar[['yas', 'gelir', 'sehir']])
```

```python
# Sayısal sütunların istatistiksel özeti
display(df.describe())

# Çarpıklık ve Basıklık
sayisal_df = df.select_dtypes(include=[np.number])
print("--- Çarpıklık ---")
print(sayisal_df.skew())
print("\n--- Basıklık ---")
print(sayisal_df.kurtosis())
```

```python
# Histogram ve Yoğunluk Eğrisi (KDE)
plt.figure(figsize=(10, 6))
sns.histplot(df['gelir'].dropna(), kde=True, bins=50, color='teal')
plt.title("Gelir Dağılımı (Histogram ve Yoğunluk Eğrisi)")
plt.xlabel("Gelir")
plt.ylabel("Frekans")
plt.show()

# Kutu Grafiği (Boxplot) ile Aykırı Değer (Outlier) Görselleştirme
plt.figure(figsize=(10, 4))
sns.boxplot(x=df['gelir'], color='coral')
plt.title("Gelir Değişkenindeki Aykırı Değerler (Boxplot)")
plt.xlabel("Gelir")
plt.show()
```

```python
# Abonelik tipleri yüzdesel dağılımı
abonelik_yuzde = df['abonelik_tipi'].value_counts(normalize=True) * 100
print("--- Abonelik Tipi Dağılımı (%) ---")
print(abonelik_yuzde)

# Churn (Terk) oranı yüzdesel dağılımı
churn_yuzde = df['churn'].value_counts(normalize=True) * 100
print("\n--- Churn Oranı (%) ---")
print(churn_yuzde)
```

```python
# Sayısal değişkenler arası korelasyon matrisi
korelasyon_matrisi = df.select_dtypes(include=[np.number]).corr()

# Korelasyon matrisinin ısı haritası (Heatmap) ile görselleştirilmesi
plt.figure(figsize=(10, 8))
sns.heatmap(korelasyon_matrisi, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1, square=True)
plt.title("Sayısal Değişkenler Arası Korelasyon Isı Haritası")
plt.show()
```
